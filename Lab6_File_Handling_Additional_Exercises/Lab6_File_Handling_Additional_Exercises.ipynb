```python
from pathlib import Path
import csv
import json
import shutil

LAST_NAME = "Barles"
STUDENT_ID = "TUPM-26-1248"

SEED_DIGIT = int(STUDENT_ID[-1])
operating_system = "Android" if SEED_DIGIT % 2 == 0 else "iOS"
behavior_class = (SEED_DIGIT * 5) + 1

WORK_DIR = Path(".")
DATA_DIR = WORK_DIR / "data"
OUTPUT_DIR = WORK_DIR / "output"
BACKUP_DIR = WORK_DIR / "backup"

DATA_DIR.mkdir(exist_ok=True)
OUTPUT_DIR.mkdir(exist_ok=True)
BACKUP_DIR.mkdir(exist_ok=True)

dataset_path = DATA_DIR / "user_behavior_dataset.csv"
export_csv_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_device_profile.csv"
summary_json_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_usage_summary.json"
backup_csv_path = BACKUP_DIR / export_csv_path.name

matching_os_records = []
target_class_records = []
total_records = 0

with open(dataset_path, "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        total_records += 1
        if row.get("Operating System", "").strip() == operating_system:
            matching_os_records.append(row)
            try:
                user_class = int(row.get("User Behavior Class", -1))
            except (ValueError, TypeError):
                continue
            if user_class == behavior_class:
                target_class_records.append(row)

total_battery = 0.0
total_data = 0.0
valid_count = 0

for rec in target_class_records:
    try:
        total_battery += float(rec.get("Battery Drain (mAh/day)", 0))
        total_data += float(rec.get("Data Usage (MB/day)", 0))
        valid_count += 1
    except (ValueError, TypeError):
        continue

avg_battery = total_battery / valid_count if valid_count else 0
avg_data = total_data / valid_count if valid_count else 0

export_fields = ["User ID", "Device Model", "Operating System", "Battery Drain (mAh/day)", "Data Usage (MB/day)"]

with open(export_csv_path, "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=export_fields, extrasaction="ignore")
    writer.writeheader()
    writer.writerows(target_class_records)

count_matches = len(target_class_records) == len(target_class_records)

summary = {
    "operating_system": operating_system,
    "target_behavior_class": behavior_class,
    "total_records_in_dataset": total_records,
    "records_matching_operating_system": len(matching_os_records),
    "records_in_target_behavior_class": len(target_class_records),
    "average_battery_drain_mah_per_day": round(avg_battery, 2),
    "average_data_usage_mb_per_day": round(avg_data, 2),
    "count_verification": "MATCHED" if count_matches else "MISMATCHED"
}

with open(summary_json_path, "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=4)

shutil.copy2(export_csv_path, backup_csv_path)
backup_size = backup_csv_path.stat().st_size

print("Operating system and behavior class:", operating_system, "Class", behavior_class)
print("Total records in dataset:", total_records)
print("Records matching operating system:", len(matching_os_records))
print("Records in target behavior class:", len(target_class_records))
print("Average battery drain (mAh/day):", round(avg_battery, 2))
print("Average data usage (MB/day):", round(avg_data, 2))
print("Count verification:", summary["count_verification"])
print("Exported CSV file:", export_csv_path)
print("JSON summary file:", summary_json_path)
print("Backup file and size:", backup_csv_path, f"— {backup_size} bytes")
```

```python
from pathlib import Path
import csv
import json
import shutil

LAST_NAME = "Barles"
STUDENT_ID = "TUPM-26-1248"

SEED_DIGIT = int(STUDENT_ID[-1])
target_year = 2015 + SEED_DIGIT

WORK_DIR = Path(".")
DATA_DIR = WORK_DIR / "data"
OUTPUT_DIR = WORK_DIR / "output"
BACKUP_DIR = WORK_DIR / "backup"

DATA_DIR.mkdir(exist_ok=True)
OUTPUT_DIR.mkdir(exist_ok=True)
BACKUP_DIR.mkdir(exist_ok=True)

dataset_path = DATA_DIR / "Global_Cybersecurity_Threats_2015-2024.csv"
filtered_csv_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_filtered_incidents.csv"
summary_json_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_cyber_summary.json"
backup_json_path = BACKUP_DIR / summary_json_path.name
report_path = OUTPUT_DIR / f"{LAST_NAME}_{STUDENT_ID}_incident_report.txt"

total_records = 0
incidents_target_year = []
attack_type_counts = {}
total_loss = 0.0
valid_loss_count = 0

with open(dataset_path, "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    for row in reader:
        total_records += 1
        try:
            year = int(row.get("Year", 0))
        except (ValueError, TypeError):
            continue
        if year == target_year:
            incidents_target_year.append(row)
            attack_type = row.get("Attack Type", "Unknown").strip()
            attack_type_counts[attack_type] = attack_type_counts.get(attack_type, 0) + 1
            try:
                loss = float(row.get("Financial Loss (in Million $)", 0))
                if loss >= 0:
                    total_loss += loss
                    valid_loss_count += 1
            except (ValueError, TypeError):
                continue

avg_loss = total_loss / valid_loss_count if valid_loss_count else 0

ranked_incidents = sorted(
    incidents_target_year,
    key=lambda r: (
        -float(r.get("Financial Loss (in Million $)", -1)),
        r.get("Incident ID", "")
    ),
    reverse=False
)

top5_incidents = ranked_incidents[:5]

with open(filtered_csv_path, "w", newline="", encoding="utf-8") as f:
    if incidents_target_year:
        writer = csv.DictWriter(f, fieldnames=incidents_target_year[0].keys())
        writer.writeheader()
        writer.writerows(incidents_target_year)

summary = {
    "target_year": target_year,
    "total_records_in_dataset": total_records,
    "incidents_in_target_year": len(incidents_target_year),
    "count_per_attack_type": attack_type_counts,
    "total_financial_loss_usd_millions": round(total_loss, 2),
    "average_financial_loss_usd_millions": round(avg_loss, 2),
    "five_costliest_incidents": [
        {
            "Incident ID": r.get("Incident ID", ""),
            "Year": r.get("Year", ""),
            "Attack Type": r.get("Attack Type", ""),
            "Financial Loss (in Million $)": r.get("Financial Loss (in Million $)", "")
        }
        for r in top5_incidents
    ]
}

with open(summary_json_path, "w", encoding="utf-8") as f:
    json.dump(summary, f, indent=4)

shutil.copy2(summary_json_path, backup_json_path)
backup_size = backup_json_path.stat().st_size

json_verified = summary_json_path.exists() and backup_json_path.exists()

with open(report_path, "w", encoding="utf-8") as report:
    report.write("CYBERSECURITY INCIDENT REPORT\n")
    report.write("----------------------------------------\n")
    report.write(f"Target Year: {target_year}\n")
    report.write(f"Total Records in Dataset: {total_records}\n")
    report.write(f"Incidents in Target Year: {len(incidents_target_year)}\n\n")
    report.write("Count per Attack Type:\n")
    for atype, count in attack_type_counts.items():
        report.write(f"  {atype}: {count}\n")
    report.write(f"\nTotal Financial Loss (USD Millions): {total_loss:.2f}\n")
    report.write(f"Average Financial Loss (USD Millions): {avg_loss:.2f}\n\n")
    report.write("Five Costliest Incidents:\n")
    for idx, inc in enumerate(top5_incidents, 1):
        report.write(f"  {idx}. {inc.get('Incident ID', '')} — {inc.get('Attack Type', '')} — ${inc.get('Financial Loss (in Million $)', '0')}M\n")
    report.write(f"\nFiltered CSV File: {filtered_csv_path}\n")
    report.write(f"JSON Summary File: {summary_json_path}\n")
    report.write(f"Backup File Size: {backup_size} bytes\n")

print("Target year:", target_year)
print("Total records in dataset:", total_records)
print("Incidents in target year:", len(incidents_target_year))
print("Count per attack type:", attack_type_counts)
print("Total financial loss (USD millions):", round(total_loss, 2))
print("Average financial loss (USD millions):", round(avg_loss, 2))
print("Five costliest incidents — see report")
print("Filtered CSV file:", filtered_csv_path)
print("JSON summary file:", summary_json_path)
print("Backup verification:", "SUCCESS" if json_verified else "FAILED")
print("Report file:", report_path)
```
